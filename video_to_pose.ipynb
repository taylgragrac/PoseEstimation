### Update the weights based on ground truth

```python
# =========================
# 0) Imports & device
# =========================
import os, json, glob, cv2
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# =========================
# 1) Get your model, unwrap, freeze backbone (we'll unfreeze head later)
#    NOTE: adjust `model.model` if your HRNet is elsewhere.
# =========================
updating_last_layer = False

if updating_last_layer:
    # your outer wrapper likely stores HRNet here:
    hrnet_model = model.model

    # move to device for probing later
    hrnet_model = hrnet_model.to(device)

    # unwrap if using DataParallel/DDP for attribute access
    if isinstance(hrnet_model, (nn.DataParallel, nn.parallel.DistributedDataParallel)):
        hrnet_base = hrnet_model.module
    else:
        hrnet_base = hrnet_model

    # freeze everything initially
    for p in hrnet_base.parameters():
        p.requires_grad = False


# =========================
# 2) Probe model to get true output heatmap shape (avoid guessing)
# =========================
if updating_last_layer:
    hrnet_model.eval()
    inW, inH = (256, 192)  # your chosen network input size (W,H)
    with torch.no_grad():
        dummy = torch.zeros(1, 3, inH, inW, device=device)
        out = hrnet_model(dummy)
        if isinstance(out, dict) and "heatmaps" in out:
            out = out["heatmaps"]
        _, J, outH, outW = out.shape
    print(f"[Probe] Output heatmaps: J={J}, size={outW}x{outH}")


# =========================
# 3) Dataset (robust image/label pairing + JSON normalizer)
# =========================
class BabyPoseDataset(Dataset):
    def __init__(self, img_dir, json_dir,
                 input_size=(256,192),      # (W, H)
                 out_size=(64,48),          # (W, H) - set from model probe
                 num_joints=17,
                 sigma=2.5,                 # slightly larger for smoother gradients
                 strip_suffixes=("_gt", "_label", "_labels", "_ann", "_pose"),
                 image_exts=(".jpg", ".jpeg", ".png", ".bmp"),
                 log_unmatched=20):
        self.img_dir = img_dir
        self.json_dir = json_dir
        self.input_size = input_size
        self.out_size = out_size
        self.num_joints = num_joints
        self.sigma = sigma
        self.strip_suffixes = strip_suffixes
        self.image_exts = image_exts
        self.log_unmatched = log_unmatched

        self.to_tensor = transforms.ToTensor()
        self.normalize = transforms.Normalize(
            mean=[0.485, 0.456, 0.406],   # ImageNet
            std=[0.229, 0.224, 0.225]
        )

        # --- Build map from normalized stem -> image path (recursive) ---
        def normalize_base(stem):
            s = stem
            for suf in strip_suffixes:
                if s.endswith(suf):
                    s = s[: -len(suf)]
            return s.lower()

        img_map = {}
        for ext in image_exts:
            for p in glob.glob(os.path.join(img_dir, "**", f"*{ext}"), recursive=True):
                stem = os.path.splitext(os.path.basename(p))[0]
                img_map.setdefault(normalize_base(stem), p)

        # --- Pair JSONs with images ---
        pairs, unmatched = [], []
        for jf in os.listdir(json_dir):
            if not jf.endswith(".json"):
                continue
            stem = os.path.splitext(jf)[0]
            norm = normalize_base(stem)
            img_path = img_map.get(norm) or img_map.get(stem.lower())
            if img_path is None:
                unmatched.append(stem)
                continue
            pairs.append((os.path.join(json_dir, jf), img_path))

        if not pairs:
            raise RuntimeError(
                f"No matching (json,image) pairs found.\n"
                f"Images under: {img_dir}\nJSONs under: {json_dir}\n"
                f"Unmatched (first {self.log_unmatched}): {unmatched[:self.log_unmatched]}"
            )
        if unmatched:
            print(f"[Dataset] Unmatched JSONs (showing up to {self.log_unmatched}): {unmatched[:self.log_unmatched]}")
            print(f"[Dataset] Total unmatched: {len(unmatched)}; matched pairs: {len(pairs)}")

        self.pairs = pairs

    # ---- Normalizer: handles nested [[ [x,y,v], ... ]] and [[x,y,v], ...] and flat [x1,y1,v1,...]
    def _normalize_kpts(self, kpts_raw, num_joints_expected=None):
        def to_vis(v):
            try:
                return 1 if float(v) > 0 else 0
            except Exception:
                return 0

        data = kpts_raw

        # auto-unbox extra single-item nesting: [ <list-of-triplets> ] -> <list-of-triplets>
        while isinstance(data, list) and len(data) == 1 and isinstance(data[0], list):
            data = data[0]

        # List of lists/tuples: [[x,y,v], ...]
        if isinstance(data, list) and data and all(isinstance(x, (list, tuple)) for x in data):
            triples = []
            for item in data:
                if len(item) >= 3 and not isinstance(item[0], (list, tuple)):
                    x, y, v = item[:3]
                    triples.append((float(x), float(y), to_vis(v)))
                elif len(item) == 2 and not isinstance(item[0], (list, tuple)):
                    x, y = item
                    triples.append((float(x), float(y), 1))
                else:
                    # Flatten one more level if needed
                    if isinstance(item, (list, tuple)) and item and isinstance(item[0], (list, tuple)):
                        for sub in item:
                            if len(sub) >= 3:
                                x, y, v = sub[:3]
                                triples.append((float(x), float(y), to_vis(v)))
                            elif len(sub) == 2:
                                x, y = sub
                                triples.append((float(x), float(y), 1))
                            else:
                                raise ValueError(f"Keypoint entry too short: {sub}")
                    else:
                        raise ValueError(f"Unexpected nesting in keypoints: {item}")
            return triples

        # Flat list fallback: [x1,y1,v1,x2,y2,v2,...]
        if isinstance(data, list) and all(isinstance(x, (int, float)) for x in data):
            if len(data) % 3 != 0:
                raise ValueError(f"Flat keypoints list length {len(data)} is not a multiple of 3.")
            triples = []
            for i in range(0, len(data), 3):
                x, y, v = data[i:i+3]
                triples.append((float(x), float(y), to_vis(v)))
            return triples

        # Empty -> []
        if isinstance(data, list) and len(data) == 0:
            return []

        raise ValueError(f"Unrecognized keypoint structure: {type(data)}")

    # ------------------------------------------------------
    def __len__(self):
        return len(self.pairs)

    def _make_gaussian(self, x, y, w, h, sigma):
        xs = np.arange(w, dtype=np.float32)
        ys = np.arange(h, dtype=np.float32)[:, None]
        return np.exp(-((xs - x)**2 + (ys - y)**2) / (2 * sigma * sigma))

    def __getitem__(self, idx):
        json_path, img_path = self.pairs[idx]

        # --- load image (original size) ---
        img_bgr = cv2.imread(img_path)
        if img_bgr is None:
            raise FileNotFoundError(f"Could not read image: {img_path}")
        H0, W0 = img_bgr.shape[:2]
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

        # --- resize to network input (W,H) ---
        inW, inH = self.input_size
        img_resized = cv2.resize(img_rgb, (inW, inH), interpolation=cv2.INTER_LINEAR)
        img = self.to_tensor(img_resized)
        img = self.normalize(img)

        # --- load & normalize keypoints (in ORIGINAL coords) ---
        with open(json_path) as f:
            raw = json.load(f)
        kpts = self._normalize_kpts(raw, num_joints_expected=self.num_joints)

        # clamp/pad to expected joint count
        if self.num_joints is not None and len(kpts) >= self.num_joints:
            kpts = kpts[:self.num_joints]
        elif self.num_joints is not None and len(kpts) < self.num_joints:
            kpts = kpts + [(0.0, 0.0, 0)] * (self.num_joints - len(kpts))

        # --- rescale to heatmap coords (from ORIGINAL -> OUTPUT heatmap size) ---
        outW, outH = self.out_size
        sx = outW / W0
        sy = outH / H0
        kpts_hm = [(kx * sx, ky * sy, 1 if v > 0 else 0) for (kx, ky, v) in kpts]

        # --- make Gaussian heatmaps ---
        heatmaps = np.zeros((self.num_joints, outH, outW), dtype=np.float32)
        for j, (x, y, vis) in enumerate(kpts_hm):
            if vis > 0 and 0 <= x < outW and 0 <= y < outH:
                heatmaps[j] = np.maximum(
                    heatmaps[j], self._make_gaussian(x, y, outW, outH, self.sigma)
                )

        heatmaps = torch.from_numpy(heatmaps)  # (J, H, W)
        return img, heatmaps


# =========================
# 4) Build dataset & dataloader (start with num_workers=0 to debug)
# =========================
if updating_last_layer:
    baby_dataset = BabyPoseDataset(
        img_dir="/content/PoseEstimation/ground_truth/images",
        json_dir="/content/PoseEstimation/ground_truth/labels",
        input_size=(inW, inH),
        out_size=(outW, outH),
        num_joints=J,
        sigma=2.5,  # matches class default
        strip_suffixes=("_gt", "_label", "_labels", "_ann", "_pose"),
    )

    baby_dataloader = DataLoader(
        baby_dataset,
        batch_size=4,
        shuffle=True,
        num_workers=4,   # bump to 2–4 once this runs cleanly
        pin_memory=True
    )


# =========================
# 5) Find & unfreeze the head, then build optimizer
# =========================
if updating_last_layer:
    # try common head names, else fallback to last Conv2d
    head = None
    for attr in ["final_layer", "head", "heads", "keypoint_head", "classifier", "fc", "last_layer"]:
        if hasattr(hrnet_base, attr):
            head = getattr(hrnet_base, attr)
            if isinstance(head, (list, tuple)):
                head = head[-1]
            break
    if head is None:
        last_conv = None
        for m in hrnet_base.modules():
            if isinstance(m, nn.Conv2d):
                last_conv = m
        if last_conv is None:
            raise RuntimeError("Could not find a head (or a Conv2d fallback) to unfreeze.")
        head = last_conv

    # unfreeze only the head
    for p in head.parameters():
        p.requires_grad = True

    # collect trainable params
    trainable_params = [p for p in hrnet_base.parameters() if p.requires_grad]
    if not trainable_params:
        raise RuntimeError("No trainable parameters found — check head detection/unfreezing.")

    print("Trainable tensors:")
    for n, p in hrnet_base.named_parameters():
        if p.requires_grad:
            print(f"  {n}  {tuple(p.shape)}")

    criterion = nn.MSELoss()
    optimizer = optim.Adam(trainable_params, lr=1e-3)  # HIGHER LR


# =========================
# 6) Train loop (with one-time shape sanity check)
# =========================
if updating_last_layer:
    hrnet_model.train()
    did_shape_check = False
    epochs = 60  # MORE EPOCHS

    for epoch in range(epochs):
        running = 0.0
        for images, gt_heatmaps in baby_dataloader:
            images = images.to(device, non_blocking=True)
            gt_heatmaps = gt_heatmaps.to(device, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)
            preds = hrnet_model(images)
            if isinstance(preds, dict) and "heatmaps" in preds:
                preds = preds["heatmaps"]

            if not did_shape_check:
                if preds.shape != gt_heatmaps.shape:
                    raise RuntimeError(
                        f"Shape mismatch: preds {tuple(preds.shape)} vs gt {tuple(gt_heatmaps.shape)}. "
                        f"Ensure out_size={(outW, outH)} and num_joints={J} match."
                    )
                did_shape_check = True

            loss = criterion(preds, gt_heatmaps)
            loss.backward()
            optimizer.step()
            running += loss.item() * images.size(0)

        print(f"Epoch {epoch}: Loss {running / len(baby_dataset):.4f}")


# =========================
# 7) Save weights (save the UNWRAPPED base to avoid 'module.' key issues)
# =========================
if updating_last_layer:
    save_path = "/content/PoseEstimation/weights/hrnet_baby.pth"
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    torch.save(hrnet_base.state_dict(), save_path)
    print(f"[Save] Weights saved to: {save_path}")


# =========================
# 8) Reload for inference with SimpleHRNet and VERIFY weights
# =========================
if updating_last_layer:
    baby_model = SimpleHRNet(
        32, J, "./weights/hrnet_baby.pth",
        multiperson=False,
        yolo_version='v8',
        yolo_model_def='yolov8n',
        device=device
    )
    print("[Load] SimpleHRNet reloaded with tuned weights.")

    # --- locate the underlying HRNet inside SimpleHRNet ---
    inner = None
    for attr in ["model", "hrnet", "pose_model", "net", "backbone"]:
        if hasattr(baby_model, attr):
            inner = getattr(baby_model, attr)
            break
    assert inner is not None, "Could not find underlying model inside SimpleHRNet."

    def head_weight_flat(m):
        base = m.module if isinstance(
            m, (nn.DataParallel, nn.parallel.DistributedDataParallel)
        ) else m
        head = getattr(base, "final_layer")
        return head.weight.detach().cpu().flatten(), head.bias.detach().cpu().flatten()

    # Compare trained-in-memory vs inside SimpleHRNet
    w_tr, b_tr = head_weight_flat(hrnet_model)  # trained weights still in memory
    w_in, b_in = head_weight_flat(inner)        # reloaded weights

    print("Diff norm (SimpleHRNet vs trained):",
          float((w_in - w_tr).norm()), float((b_in - b_tr).norm()))
```
